Logistic Regression

```python
# Import libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
# Load dataset
df = sns.load_dataset('tips')

print("===== DATASET PREVIEW =====")
print(df.head())

# EDA

print("\n===== DATA INFO =====")
print(df.info())

print("\n===== STATISTICAL SUMMARY =====")
print(df.describe())

print("\n===== MISSING VALUES =====")
print(df.isnull().sum())
```

VISUALIZATIONS

```python
# Histogram
plt.figure()
df['total_bill'].hist()
plt.title("Distribution of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()

# Scatter Plot
plt.figure()
plt.scatter(df['total_bill'], df['tip'])
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.title("Total Bill vs Tip")
plt.show()

# Box Plot
plt.figure()
df.boxplot(column='total_bill')
plt.title("Boxplot of Total Bill")
plt.show()
```

```python
# CREATE TARGET VARIABLE

df['HighTip'] = df['tip'].apply(lambda x: 1 if x > 3 else 0)
```

```python
# MODEL BUILDING

X = df[['total_bill', 'size']]
y = df['HighTip']

model = LogisticRegression()
model.fit(X, y)
```

```python
# PREDICTION

new_data = [[20, 3]]
prob = model.predict_proba(new_data)

print("\n===== PREDICTION =====")
print("Total Bill = 20, Size = 3")
print("Probability of High Tip:", prob[0][1])
```
